# Probability Calculations & Naive Bayes Concepts

This notebook demonstrates calculating marginal, joint, and conditional probabilities on categorical weather datasets.

### 1. Import Pandas

```python
import pandas as pd
import numpy as np
```

### 2. Create Sample Weather Table

```python
weather_data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rainy', 'Rainy', 'Rainy', 'Overcast'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(weather_data)
df
```

### 3. Calculate Marginal Probabilities

```python
total_rows = len(df)
yes_count = len(df[df['Play'] == 'Yes'])
no_count = len(df[df['Play'] == 'No'])

print(f'P(Play = Yes) = {yes_count}/{total_rows} = {yes_count/total_rows:.4f}')
print(f'P(Play = No)  = {no_count}/{total_rows} = {no_count/total_rows:.4f}')
```

### 4. Calculate Conditional Probabilities P(Outlook | Play)

```python
overcast_yes = len(df[(df['Outlook'] == 'Overcast') & (df['Play'] == 'Yes')])
print(f'P(Outlook=Overcast | Play=Yes) = {overcast_yes}/{yes_count} = {overcast_yes/yes_count:.4f}')
```
