# K-Nearest Neighbors (KNN) — Student Result Prediction Task

This notebook demonstrates implementing the **KNN Classifier** from `scikit-learn` to predict student Pass/Fail exam outcomes based on study hours and attendance.

### 1. Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

### 2. Create Training Dataset

```python
data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Hours': [1, 2, 3, 5, 6, 7],
    'Attendance': [50, 55, 60, 75, 80, 90],
    'Result': ['Fail', 'Fail', 'Fail', 'Pass', 'Pass', 'Pass']
}
df = pd.DataFrame(data)
df
```

### 3. Separate Features (X) and Target (y)

```python
X = df[['Hours', 'Attendance']]  # Independent features
y = df['Result']                # Target label
print('Features Matrix (X):\n', X)
print('\nTarget Vector (y):\n', y)
```

### 4. Train KNN Model (`n_neighbors=3`)

```python
# Instantiate KNN Classifier with k=3
knn = KNeighborsClassifier(n_neighbors=3)

# Train the model using fit()
knn.fit(X, y)
print('KNN Model trained successfully with k=3!')
```

### 5. Predict Outcome for Unseen Test Sample

```python
# Test sample: Student with 5 study hours and 78% attendance
new_student = [[5, 78]]
predicted_result = knn.predict(new_student)
print(f'Prediction for Hours=5, Attendance=78% -> {predicted_result[0]}')
```
