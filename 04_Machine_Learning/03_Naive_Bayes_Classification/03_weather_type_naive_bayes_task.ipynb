# Naive Bayes Classification — Weather Type Classification Task

This notebook demonstrates data preprocessing, one-hot encoding (`pd.get_dummies`), target label encoding, and training a **Gaussian / Multinomial Naive Bayes Classifier**.

### 1. Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

### 2. Load Weather Dataset (`weather.csv`)

```python
weather_csv = os.path.join('..', '..', '03_Data_Analysis_Tasks', '09_Weather_Data_EDA', 'weather.csv')
if not os.path.exists(weather_csv):
    weather_csv = 'weather.csv'
df = pd.read_csv(weather_csv)
df.head()
```

### 3. Missing Values Check

```python
print('Missing Values:\n', df.isna().sum())
```

### 4. Separate Features Matrix (X) & Target (y)

```python
X = df.drop('Weather Type', axis=1)
y = df['Weather Type']
print('Features Shape:', X.shape)
print('Target Shape:', y.shape)
```

### 5. One-Hot Encode Categorical Features (`pd.get_dummies`)

```python
X_encoded = pd.get_dummies(X)
print('Encoded Features Shape:', X_encoded.shape)
X_encoded.head()
```

### 6. Encode Categorical Target Labels

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
print('Target Classes:', label_encoder.classes_)
```

### 7. Train-Test Split (80% Train, 20% Test)

```python
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y_encoded, test_size=0.2, random_state=42)
print('X_train Shape:', X_train.shape)
print('X_test Shape:', X_test.shape)
```

### 8. Train Naive Bayes Classifier

```python
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
y_pred = nb_model.predict(X_test)
```

### 9. Evaluate Classification Accuracy

```python
acc = accuracy_score(y_test, y_pred)
print(f'Naive Bayes Model Accuracy: {acc * 100:.2f}%')
print('\nClassification Report:\n', classification_report(y_test, y_pred))
```
