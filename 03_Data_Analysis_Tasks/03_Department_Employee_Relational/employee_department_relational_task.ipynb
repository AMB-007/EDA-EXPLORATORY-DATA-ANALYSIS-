## Level 5 — Multiple Aggregations

### Q1: Department salary metrics (Average, Maximum, Minimum)

```python
e_df.groupby('department')['salary'].agg(['mean', 'max', 'min'])
```

### Q2: Department metrics (Avg Performance, Avg Satisfaction, Avg Attendance)

```python
e_df.groupby('department').agg({
    'performance_score': 'mean',
    'satisfaction_score': 'mean',
    'attendance_percentage': 'mean'
})
```

### Q3: Work Mode metrics (Employee Count, Avg Salary, Avg Performance)

```python
e_df.groupby('work_mode').agg(
    employee_count=('employee_id', 'count'),
    avg_salary=('salary', 'mean'),
    avg_performance=('performance_score', 'mean')
)
```

### Q4: Education Level metrics (Avg Salary, Avg Experience, Avg Performance)

```python
e_df.groupby('education').agg({
    'salary': 'mean',
    'experience_years': 'mean',
    'performance_score': 'mean'
})
```

## Challenge Questions

### C1: Find the department with the highest average salary

```python
top_dept_salary = e_df.groupby('department')['salary'].mean()
print('Department:', top_dept_salary.idxmax())
print('Highest Avg Salary: $', top_dept_salary.max())
```

### C2: Find the department with the highest average performance score

```python
top_dept_perf = e_df.groupby('department')['performance_score'].mean()
print('Department:', top_dept_perf.idxmax())
print('Highest Avg Performance:', top_dept_perf.max())
```

### C3: Find the city having the largest number of employees

```python
top_city = e_df['city'].value_counts()
print('City:', top_city.idxmax())
print('Employee Count:', top_city.max())
```

### C4: Find the employee with the highest total compensation (salary + bonus)

```python
e_df_comp = e_df.copy()
e_df_comp['total_compensation'] = e_df_comp['salary'] + e_df_comp['bonus']
e_df_comp.nlargest(1, 'total_compensation')[['first_name', 'department', 'salary', 'bonus', 'total_compensation']]
```

### C5: Find employees whose salary is above their department's average salary

```python
dept_avg_sal = e_df.groupby('department')['salary'].transform('mean')
above_avg_emp = e_df[e_df['salary'] > dept_avg_sal]
print('Count of employees earning above department average:', len(above_avg_emp))
above_avg_emp[['first_name', 'department', 'salary']].head()
```

### C6: Find the top 3 highest-paid employees from each department

```python
top3_per_dept = e_df.groupby('department', group_keys=False).apply(
    lambda g: g.nlargest(3, 'salary')
)[['department', 'first_name', 'salary']]
top3_per_dept
```
