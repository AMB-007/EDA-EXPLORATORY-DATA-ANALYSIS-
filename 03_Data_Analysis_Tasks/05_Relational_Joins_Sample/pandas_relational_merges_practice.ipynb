# Pandas Merges & Relational Joins Practice

This notebook demonstrates joining relational customer (`custom_windows.csv`) and order (`order_windows.csv`) tables using `pd.merge()` with `inner`, `left`, `right`, and `outer` joins, as well as sorting and filtering merged DataFrames.

### 1. Import Pandas

```python
import pandas as pd
```

### 2. Load Customer & Order Datasets

```python
# Load customer dataset
c_data = pd.read_csv('custom_windows.csv')
c_df = pd.DataFrame(c_data)
print('=== Customers Table ===')
display(c_df)

# Load order dataset
o_data = pd.read_csv('order_windows.csv')
o_df = pd.DataFrame(o_data)
print('\n=== Orders Table ===')
display(o_df)
```

### 3. Merging Tables (`pd.merge`)

#### 3.1 Inner Join (`how='inner'`)

```python
# Inner Join: Returns matching records in both tables
result_inner = pd.merge(c_df, o_df, on='id', how='inner')
print('=== Inner Join ===')
display(result_inner)
```

#### 3.2 Left Join (`how='left'`)

```python
# Left Join: Returns all customers and matched order records
result_left = pd.merge(c_df, o_df, on='id', how='left')
print('=== Left Join ===')
display(result_left)
```

#### 3.3 Right Join (`how='right'`)

```python
# Right Join: Returns all orders and matched customer records
result_right = pd.merge(c_df, o_df, on='id', how='right')
print('=== Right Join ===')
display(result_right)
```

#### 3.4 Outer Join (`how='outer'`)

```python
# Outer Join: Returns all records from both tables
result_outer = pd.merge(c_df, o_df, on='id', how='outer')
print('=== Outer Join ===')
display(result_outer)
```

### 4. Sorting Merged Data

#### 4.1 Sort Merged Table by Order Amount (Descending)

```python
# Inner join sorted by amount descending
result_desc = pd.merge(c_df, o_df, on='id', how='inner').sort_values(by='amount', ascending=False)
result_desc
```

#### 4.2 Sort Merged Table by Order Amount (Ascending)

```python
# Inner join sorted by amount ascending
result_asc = pd.merge(c_df, o_df, on='id', how='inner').sort_values(by='amount', ascending=True)
result_asc
```

### 5. Filtering Merged Results

#### 5.1 Filter Customers with Salary > 2000

```python
# Filter merged inner join where customer salary > 2000
result_filtered = result_inner.loc[result_inner['salary'] > 2000]
result_filtered
```
