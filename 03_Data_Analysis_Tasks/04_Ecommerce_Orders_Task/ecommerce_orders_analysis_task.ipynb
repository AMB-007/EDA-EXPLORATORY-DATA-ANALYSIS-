# E-Commerce Orders & Customers Data Analysis Task (50 Questions)

This notebook contains complete solutions for the **50 Student Practice Questions** covering single dataset exploration, joining orders and customer tables, conditional filtering, and group aggregations.

## Setup & Data Loading

```python
import pandas as pd
import numpy as np

# Load datasets
orders = pd.read_csv("orders.csv")
customers = pd.read_csv("customers.csv")

print("Orders Shape:", orders.shape)
print("Customers Shape:", customers.shape)
orders.head(3)
```

### Q15: Sort DataFrame by TotalAmount from highest to lowest

```python
orders.sort_values('TotalAmount', ascending=False)
```

### Q16: Display top 5 highest-value orders

```python
orders.nlargest(5, 'TotalAmount')
```

### Q17: Find product with highest TotalAmount

```python
max_idx = orders['TotalAmount'].idxmax()
orders.loc[max_idx, ['Product', 'Category', 'TotalAmount']]
```

### Q18: Count orders placed in each City and Status combination

```python
orders.groupby(['City', 'Status']).size()
```

### Q19: Find total revenue from Completed orders only

```python
completed_rev = orders[orders['Status'] == 'Completed']['TotalAmount'].sum()
print('Completed Orders Revenue: $', completed_rev)
```

```python
city_rev = orders.groupby('City')['TotalAmount'].sum()
city_rev[city_rev > 20000]
```

### Q21: Column PriceLevel: 'High' if UnitPrice >= 10000 else 'Normal'

```python
orders['PriceLevel'] = np.where(orders['UnitPrice'] >= 10000, 'High', 'Normal')
orders[['Product', 'UnitPrice', 'PriceLevel']]
```

### Q22: Column OrderValueLevel: 'Large' if TotalAmount >= 15000 else 'Regular'

```python
orders['OrderValueLevel'] = np.where(orders['TotalAmount'] >= 15000, 'Large', 'Regular')
orders[['Product', 'TotalAmount', 'OrderValueLevel']]
```

### Q23: Find average Quantity for Electronics and Furniture separately

```python
orders[orders['Category'].isin(['Electronics', 'Furniture'])].groupby('Category')['Quantity'].mean()
```

### Q24: Find category with highest total revenue

```python
top_cat_rev = orders.groupby('Category')['TotalAmount'].sum()
print('Category:', top_cat_rev.idxmax())
print('Total Revenue: $', top_cat_rev.max())
```

### Q25: Groupby multiple aggregations for Category

```python
orders.groupby('Category').agg(
    Total_Quantity=('Quantity', 'sum'),
    Avg_UnitPrice=('UnitPrice', 'mean'),
    Total_Revenue=('TotalAmount', 'sum')
)
```

## Part B — Joining Orders and Customers (Q26 to Q35)

### Q26: Load customers.csv into second DataFrame

```python
customers = pd.read_csv('customers.csv')
customers.head()
```

### Q27: Perform inner join between orders and customers on CustomerID

```python
merged = pd.merge(orders, customers, on='CustomerID', suffixes=('_order', '_customer'))
merged.head()
```

### Q28: Select and display specified columns after join

```python
merged[['CustomerName', 'Product', 'Quantity', 'TotalAmount', 'Membership', 'City_customer']]
```

### Q29: Find total revenue generated by each customer

```python
merged.groupby('CustomerName')['TotalAmount'].sum()
```

### Q30: Find total revenue generated by each Membership type

```python
merged.groupby('Membership')['TotalAmount'].sum()
```

### Q31: Find average order value for each Membership type

```python
merged.groupby('Membership')['TotalAmount'].mean()
```

### Q32: Find total quantity purchased by customers from each City

```python
merged.groupby('City_customer')['Quantity'].sum()
```

### Q33: Find number of orders made by each Gender

```python
merged.groupby('Gender')['OrderID'].count()
```

### Q34: Find average customer age for each Membership type

```python
merged.groupby('Membership')['Age'].mean()
```

### Q35: Find customer who generated highest total revenue

```python
top_cust = merged.groupby('CustomerName')['TotalAmount'].sum()
print('Customer:', top_cust.idxmax())
print('Total Spending: $', top_cust.max())
```

## Part C — Conditions + Join (Q36 to Q50)

### Q36: Display Gold members who placed Completed orders

```python
merged[(merged['Membership'] == 'Gold') & (merged['Status'] == 'Completed')]
```

### Q37: Find Gold members whose TotalAmount > 10,000

```python
merged[(merged['Membership'] == 'Gold') & (merged['TotalAmount'] > 10000)]
```

### Q38: Find customers older than 30 who placed Completed orders

```python
merged[(merged['Age'] > 30) & (merged['Status'] == 'Completed')]
```

### Q39: Find Female customers from Kochi and display order details

```python
merged[(merged['Gender'] == 'Female') & (merged['City_customer'] == 'Kochi')]
```

### Q40: Find Male customers who purchased Electronics products

```python
merged[(merged['Gender'] == 'Male') & (merged['Category'] == 'Electronics')]
```

### Q41: Find Silver members whose Quantity > 1

```python
merged[(merged['Membership'] == 'Silver') & (merged['Quantity'] > 1)]
```

### Q42: Find customers from Thrissur with TotalAmount > 5000

```python
merged[(merged['City_customer'] == 'Thrissur') & (merged['TotalAmount'] > 5000)]
```

### Q43: Gold members whose total spending > 15,000 using groupby()

```python
gold_spend = merged[merged['Membership'] == 'Gold'].groupby('CustomerName')['TotalAmount'].sum()
gold_spend[gold_spend > 15000]
```

### Q44: Total revenue of Completed orders for each Membership type

```python
merged[merged['Status'] == 'Completed'].groupby('Membership')['TotalAmount'].sum()
```

### Q45: Total revenue generated by each City for Gold members only

```python
merged[merged['Membership'] == 'Gold'].groupby('City_customer')['TotalAmount'].sum()
```

### Q46: Average TotalAmount for Male and Female customers separately

```python
merged.groupby('Gender')['TotalAmount'].mean()
```

### Q47: Find customers who have at least one Cancelled order

```python
cancelled_custs = merged[merged['Status'] == 'Cancelled']['CustomerName'].unique()
print('Customers with Cancelled Orders:', cancelled_custs)
```

### Q48: Find customers who purchased Furniture and have Gold membership

```python
merged[(merged['Category'] == 'Furniture') & (merged['Membership'] == 'Gold')]
```

### Q49: Top 3 customers based on total Completed-order revenue

```python
merged[merged['Status'] == 'Completed'].groupby('CustomerName')['TotalAmount'].sum().nlargest(3)
```

### Q50: Create final customer summary table

```python
summary = merged.groupby('CustomerName').agg(
    Membership=('Membership', 'first'),
    City=('City_customer', 'first'),
    Order_Count=('OrderID', 'count'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Revenue=('TotalAmount', 'sum')
).reset_index()
summary
```
