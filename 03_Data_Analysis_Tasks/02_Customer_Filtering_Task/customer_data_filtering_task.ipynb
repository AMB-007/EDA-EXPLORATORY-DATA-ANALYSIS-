# Customer Dataset - Exploratory Data Analysis & Indexing Task

This notebook demonstrates data inspection, label/positional indexing (`loc`/`iloc`), conditional filtering, data sorting, and group-by aggregations on customer records.

### 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
```

### 2. Load Dataset (`customer5_windows.csv`)

```python
# Load customer data into DataFrame
data = pd.read_csv('customer5_windows.csv')
df = pd.DataFrame(data)
df.head()
```

### 3. Data Inspection Commands

#### 3.1 Display First 10 Rows (`iloc[0:10]`)

```python
# Extract first 10 rows using positional indexing
df.iloc[0:10]
```

#### 3.2 Display Last 10 Rows (`iloc[-10:, :]`)

```python
# Extract last 10 rows using positional indexing
df.iloc[-10:, :]
```

#### 3.3 Random Sample of 10 Rows (`df.sample(10)`)

```python
# Display 10 random rows from dataset
df.sample(10)
```

#### 3.4 Inspect Dataset Dimensions (`df.shape`)

```python
# Returns tuple of (number of rows, number of columns)
df.shape
```

#### 3.5 Display Column Names (`df.columns`)

```python
# List all column headers
df.columns
```

#### 3.6 Inspect Data Types (`df.dtypes`)

```python
# Check data type of each column
df.dtypes
```

#### 3.7 Overall DataFrame Information (`df.info()`)

```python
# Display memory usage, non-null counts, and data types
df.info()
```

#### 3.8 Summary Statistics (`df.describe()`)

```python
# Generate descriptive statistics for numerical columns
df.describe()
```

#### 3.9 Unique Professions & Locations

```python
# Distinct values in 'prof' column
print('Unique Professions:', df['prof'].unique())

# Distinct values in 'loc' column
print('Unique Locations:', df['loc'].unique())
```

### 4. Column Selection & Conditional Filtering

#### 4.1 Selecting Specific Columns (`fname`, `lname`, `salary`)

```python
# Subset DataFrame with selected columns
df[['fname', 'lname', 'salary']].head()
```

#### 4.2 Filter High Salary Customers (`salary > 100,000`)

```python
# Customers earning more than 100k
df[df['salary'] > 100000].head()
```

#### 4.3 Filter Customers Older than 50 (`age > 50`)

```python
# Customers age > 50
df[df['age'] > 50].head()
```

#### 4.4 Filter Customers by Profession (`prof == 'Pilot'`)

```python
# Filter customers whose profession is Pilot
df[df['prof'] == 'Pilot'].head()
```

#### 4.5 Filter Customers by Location (`loc == 'india'`)

```python
# Filter customers located in India
df[df['loc'] == 'india'].head()
```

#### 4.6 Multi-condition Filtering: Salary Range (50k to 100k)

```python
# Filter salary between 50,000 and 100,000 using logical AND (&)
df.loc[(df['salary'] >= 50000) & (df['salary'] <= 100000)].head()
```

#### 4.7 Multi-condition Filtering: Young High Earners (`age < 30` & `salary > 50k`)

```python
# Filter age < 30 and salary > 50,000
df.loc[(df['age'] < 30) & (df['salary'] > 50000)].head()
```

#### 4.8 Logical OR Filtering: Location India or UK (`loc == 'india'` | `loc == 'uk'`)

```python
# Filter customers located in either India or UK
df.loc[(df['loc'] == 'india') | (df['loc'] == 'uk')].head()
```

#### 4.9 String Substring Matching: Profession contains 'Computer'

```python
# Search string in profession column using str.contains()
df.loc[(df['prof'].str.contains('Computer', na=False)) & (df['salary'] > 50000)].head()
```

#### 4.10 Exclusion Filter: Profession is NOT 'Pilot'

```python
# Exclude Pilots using not equal operator (!=)
df.loc[df['prof'] != 'Pilot'].head()
```

### 5. Advanced Slicing & Conditional Updates (`iloc` & `loc`)

#### 5.1 Sub-grid Selection (`iloc[:10, :5]`)

```python
# Select first 10 rows and first 5 columns by position
df.iloc[:10, :5]
```

#### 5.2 Row Range Selection (`iloc[20:31]`)

```python
# Select rows 20 to 30 by position
df.iloc[20:31]
```

#### 5.3 Specific Column Indices (`iloc[:, [1, 4, 6]]`)

```python
# Select all rows for column positions 1, 4, and 6
df.iloc[:, [1, 4, 6]].head()
```

#### 5.4 Label Slicing with Specific Output Columns (`loc`)

```python
# Select fname, lname, salary for customers earning > 100k
df.loc[df['salary'] > 100000, ['fname', 'lname', 'salary']].head()
```

#### 5.5 Conditional Selection: India & Age > 40

```python
# Filter customers in India older than 40
df.loc[(df['loc'] == 'india') & (df['age'] > 40)].head()
```

#### 5.6 Conditional Value Update: Salary Bonus for Low Earners

```python
# Increment salary by 5000 for customers earning under 30,000
df.loc[df['salary'] < 30000, 'salary'] += 5000
print('Updated low salary counts:', len(df[df['salary'] < 30000]))
```

#### 5.7 String Clean-up: Capitalize Location for Pilots

```python
# Update 'india' to 'India' for Pilots
df.loc[(df['prof'] == 'Pilot') & (df['loc'] == 'india'), 'loc'] = 'India'
```

### 6. Data Sorting & Summary Aggregations

#### 6.1 Sort DataFrame by Salary (Ascending)

```python
# Sort values by salary ascending
df.sort_values('salary').head()
```

#### 6.2 Top 10 Highest Earners (`sort_values` Descending)

```python
# Sort by salary descending and return top 10
df.sort_values('salary', ascending=False).head(10)
```

#### 6.3 Multi-column Sorting (Location Ascending, Salary Descending)

```python
# Sort by location A-Z, then salary high-to-low
df.sort_values(['loc', 'salary'], ascending=[True, False]).head(10)
```

#### 6.4 Key Column Aggregations (Mean, Min, Max, Count)

```python
print('Mean Salary:', df['salary'].mean())
print('Max Salary:', df['salary'].max())
print('Min Salary:', df['salary'].min())
print('Mean Age:', df['age'].mean())
print('\nLocation Counts:\n', df['loc'].value_counts())
```

### 7. Groupby Aggregations

#### 7.1 Average Salary by Profession

```python
# Group by profession and calculate mean salary
df.groupby('prof')['salary'].mean()
```

#### 7.2 Maximum Salary by Location

```python
# Group by location and find max salary
df.groupby('loc')['salary'].max()
```

#### 7.3 Minimum Salary by Profession

```python
# Group by profession and find min salary
df.groupby('prof')['salary'].min()
```

#### 7.4 Count of Customers per Profession

```python
# Number of records per profession
df.groupby('prof').size()
```

#### 7.5 Average Salary by Location (Sorted High to Low)

```python
# Group by location, calculate mean salary, sort descending
df.groupby('loc')['salary'].mean().sort_values(ascending=False)
```

#### 7.6 Multiple Aggregations (`agg({'salary': 'mean', 'age': 'mean'})`)

```python
# Compute average salary and average age per profession
df.groupby('prof').agg({'salary': 'mean', 'age': 'mean'})
```
